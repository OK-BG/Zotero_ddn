```python
plt.figure(figsize=(11, 7))
sns.set_theme(style="whitegrid")

# Couleurs et formes uniquement pour les deux outils d'origine
couleurs_outils = {"RFDiffusion": "#1f77b4", "BindCraft": "#ff7f0e"}
styles_outils = {"RFDiffusion": "o", "BindCraft": "X"}

# Tracé des données réelles
plot = sns.scatterplot(
    data=df_clean,
    x="e-value",
    y="similarite_pourcent",
    hue="Outil",
    style="Outil",
    markers=styles_outils,
    palette=couleurs_outils,
    s=200,
    edgecolor="black",
    linewidth=1.2
)

# Passage de l'axe X en échelle logarithmique pour étirer la zone critique
plt.xscale('log')

# Repères bioinformatiques (Seuil de significativité et fenêtre d'identité cible)
plt.axvline(x=1e-5, color="red", linestyle="--", linewidth=1.5, alpha=0.8, label="Seuil strict (E-value = 1e-5)")
plt.axhspan(20, 35, color="green", alpha=0.1, label="Zone cible de sécurité (20-35%)")

# Titres et labels des axes
plt.title("Résultats PSI-BLAST : Comparaison de l'identité de séquence avec le meilleur hit humain de chaque binder", fontsize=13, fontweight='bold', pad=15)
plt.xlabel("E-value brute (Échelle Logarithmique — Zone critique à gauche)", fontsize=11, labelpad=10)
plt.ylabel("Pourcentage d'identité locale (%)", fontsize=11, labelpad=10)

# Ajustement automatique des limites de l'axe X basé sur tes vrais min/max
plt.xlim(df_clean["e-value"].min() * 0.1, df_clean["e-value"].max() * 10)
plt.ylim(15, 60)

# Graduation personnalisée pour l'axe X pour une lecture rapide
ticks = [1e-6, 1e-5, 1e-3, 1e-1, 1.0, 5.0]
labels = ["1e-6", "1e-5 (Seuil)", "0.001", "0.1", "1.0", "5.0"]
plt.xticks(ticks, labels)

# Légende calée proprement en bas à droite
plt.legend(loc='upper right', fontsize=11, frameon=True, shadow=True)

plt.tight_layout()
plt.show()
```
